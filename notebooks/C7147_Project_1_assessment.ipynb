## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
  sequence = sequence.upper()                                                     #puts all characters of sequence in upper case
  errors = []                                                                     #opens empty list
  if len(sequence) != len(reference):                                             #if the length of sequence does not equal length of reference, "wrong_length" will be printed
        errors.append("wrong_length")
  if not all(base in "ACGT" for base in sequence):                                #if the characters in the sequence are not all "ACGT", "invalid_characters" will be printed
        errors.append("invalid_characters")

  return {                                                                        #open dictionary
      "sequence": sequence,                                                       #storing uppercase sequence
      "valid": len(errors) == 0,                                                  # True if no errors found
      "errors": errors}                                                           #contains any errors found" is this right and does it come ONLY from the worksheets
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
  mutations = []                                                                 # Create empty list

  for i, base in enumerate(sequence):                                            # Loop through each option
      if base != reference[i]:                                                   # Check if each base is not equal to reference
        mutations.append({                                                       # Adds dictionary describing each mutation
              "position": i + 1,                                                #
              "reference_base": reference[i],                                    #
              "read_base": base                                                  #
        })
  return mutations                                                               # Returns list of mutations, empty if none found
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    analysed_reads = []                                                          # Create empty list

    for read in reads:                                                           # Loop through each read in the dictionary
      validation = validate_read(read["sequence"], reference)                    # Uses function from step 1

      if validation ["valid"] == True:                                           # If sequence passed validation
          mutations = find_mutations(validation["sequence"], reference)          # Check differences from reference
          status = "valid"                                                       # Set status to valid
      else:                                                                      # If not,
          mutations = []                                                         # Creates empty list
          status = "invalid"                                                     # Set status to invalid

      result = {                                                                 # Create dictionary for analysed sequence
          "read_id": read["read_id"],                                            # Store original read ID
          "sample": read["sample"],                                              # Store original sample name
          "group": read["group"],                                                # Store original experimental group
          "sequence": validation["sequence"],                                    # Store Normalised upper-case sequence
          "status": status,                                                      # Store "Valid" or "Invalid"
          "errors": validation["errors"],                                        # Store validation errors
          "mutations": mutations                                                 # Store mutations found
      }

      analysed_reads.append(result)                                             # Add this to list

    return analysed_reads                                                        # Return all analysed reads
````

## Step 4 — Summarise mismatch patterns between samples

## Function: `summarise_samples(analysed_reads)`

### 👍 You can assume:

### 🎯 Aims:

**Table of summarised results**

Return the results in the following structure, with one nested dictionary per sample, e.g. (this is an example of the format, not a correct numerical output):

### 💻 Enter your code here

````python
# Write your code here.

def summarise_samples(analysed_reads):
    sample_summary = {}                                                             # Create empty dictionary for samples

    for read in analysed_reads:                                                     # Loop through each analysed read
        sample = read["sample"]                                                     # For sample name

        if sample not in sample_summary:                                            # Check if sample is new
           sample_summary[sample] = {                                               # Create dictionary for sample
              "group": read["group"],                                               # Store group
              "valid_reads": 0,                                                     # Start count of valid reads at 0
              "invalid_reads": 0,                                                   # Start count of invalid reads at 0
              "mismatch_reads": 0,                                                  # Count valid reads which contain mutations
              "mismatch_percentage": None,                                          # Store percentage of valid reads with mutations
              "mismatch_events": 0,                                                 # Count total individual mutations
              "substitution_counts": {},                                            # Store counts of each substitution
              "position_counts": {},                                                # Store mutation counts at each position
              "top_positions": []                                                   # Store three most frequently mutated positions

           }
        if read["status"] == "valid":                                               # Check if read is valid
          sample_summary[sample]["valid_reads"] += 1                                # Increase count of valid reads

          if len(read["mutations"]) > 0:                                            # Check if read has mutations
            sample_summary[sample]["mismatch_reads"] += 1                           # Count the reads with mutations

          sample_summary[sample]["mismatch_events"] += len(read["mutations"])       # Add number of mismatches

          for mutation in read["mutations"]:                                        # Loop through each mutation
            substitution = mutation["reference_base"] + ">" + mutation["read_base"] # Create substitution name

            if substitution not in sample_summary[sample]["substitution_counts"]:   # If substitution is new
              sample_summary[sample]["substitution_counts"][substitution] = 0       # Start count at 0

            sample_summary[sample]["substitution_counts"][substitution] += 1        # Count substitution

            position = mutation["position"]                                         # Get mismatch position

            if position not in sample_summary[sample]["position_counts"]:           # If position is new
               sample_summary[sample]["position_counts"][position] = 0              # Start count at 0

            sample_summary[sample]["position_counts"][position] += 1                # Count position

        else:                                                                       # If read invalid
          sample_summary[sample]["invalid_reads"] += 1                              # Increase count of invalid reads by 1

    for sample in sample_summary:                                                   # Loop through each sample
        if sample_summary[sample]["valid_reads"] > 0:                               # Check sample is valid
           sample_summary[sample]["mismatch_percentage"] = (                        # Calculate percentage of mismatches
              sample_summary[sample]["mismatch_reads"] /                            # Number of mismatch reads divided by
              sample_summary[sample]["valid_reads"] * 100)                          # Valid reads and multiplied by 100

        else:                                                                       # If no valid reads
            sample_summary[sample]["mismatch_percentage"] = None                    # Cannot calculate when no valid reads given

        position_counts = sample_summary[sample]["position_counts"]                 # Get dictionary of mismatch positions

        ranked_positions = sorted(position_counts)                                  # Sort positions from lowest to highest

        ranked_positions = sorted(                                                  # Sort positions from highest to lowest
            ranked_positions,                                                       # Use sorted positions from before
            key=lambda position: position_counts[position],                         # Take each position one at a time and find how many mismatches occur at position
            reverse=True                                                            # Sort from highest to lowest
        )

        sample_summary[sample]["top_positions"] = ranked_positions[:3]              # Create list for top 3 positions



    groups = []                                                                     # Create empty list of group names

    for sample in sample_summary:                                                   # Loop through each sample
        group = sample_summary[sample]["group"]                                     # Get group name for sample

        if group not in groups:                                                     # Check whether group has already been added
            groups.append(group)                                                    # Add to list

    for group in groups:                                                            # Loop through each group
        print(f"Group: {group}")                                                    # Print group name

        print("Table of Summarised Results")                                        # Print title of first table
        print(
            f"{'Sample': <15}"
            f"{'Valid': >10}"
            f"{'Invalid': >10}"
            f"{'Mismatch %': >15}"
            f"{'Events': >10}"
            f"{'Top Positions': >20}"                                              # Print headings
        )
        for sample in sample_summary:                                               # Loop through each sample
            if sample_summary[sample]["group"] == group:                            # Check group name matches
                result = sample_summary[sample]                                     # Get stored results for this sample

                if result ["mismatch_percentage"] == None:                          # Check if percentage cannot be calculated
                    percentage = "N/A"                                              # Display N/A when no valid reads found
                else:                                                               # If percentage already calculated
                  percentage = f"{result['mismatch_percentage']: .1f}"              # Format to one decimal place


                print(                                                              # Print sample results
                    f"{sample}"                                                     # Sample name
                    f"{result['valid_reads']: >10}"                                 # Print number of valid reads
                    f"{result['invalid_reads']: >10}"                               # Print number of invalid reads
                    f"{percentage: >15}"                                            # Print percentage of mismatches
                    f"{result['mismatch_events']: >10}"                             # Print number of mismatches
                    f"{str(result['top_positions']): >20}"                          # Print top three mismatches

                )

        print("Table of Substitution Mutations")                                    # Print title of second table
        print("Counts and percentages per 100 valid reads")                         # Values in second table

        group_samples = []                                                          # Create list of samples in group

        for sample in sample_summary:                                               # Loop through each sample
            if sample_summary[sample]["group"] == group:                            # Check group name matches
                group_samples.append(sample)                                        # Add sample to list

        substitutions = []                                                          # List of observed substitutions

        for sample in group_samples:                                                # Loop through each sample
            for substitution in sample_summary[sample]["substitution_counts"]:      # Loop through each substitution
                if substitution not in substitutions:                               # Check substitution is new
                    substitutions.append(substitution)                              # Add to list



        substitutions = sorted(substitutions)                                       # Sort substitutions alphabetically

        print(f"{'Substitutions':<15}", end="")                                     # Print first column heading

        for sample in group_samples:                                                # Loop through samples in this group
            print(f"{sample:>24}", end="")                                          # Print sample name as column heading

        print()                                                                     # Move to next line after headings

        for substitution in substitutions:                                          # Loop through each substitution type
            print(f"{substitution:<15}", end="")                                    # Print substitution name at start of row

            for sample in group_samples:                                            # Loop through each sample
                counts = sample_summary[sample]["substitution_counts"]              # Get substitution counts
                valid_reads = sample_summary[sample]["valid_reads"]                 # Get number of valid reads

                if substitution in counts:                                          # Check whether substitution was observed
                    count = counts[substitution]                                    # Get number of occurrences
                else:                                                               # If substitution was not observed
                    count = 0                                                       # Set count to zero

                if valid_reads > 0:                                                 # Check whether sample has valid reads
                    percentage = count / valid_reads * 100                          # Calculate events per 100 valid reads
                    results = f"{count} ({percentage:.1f}%)"                        # Combine count and percentage
                else:                                                               # If there are no valid reads
                    results = f"{count} (N/A)"                                      # Percentage cannot be calculated

                print(f"{results:>24}", end="")                                     # Print value in sample column

            print()                                                                 # Move to next row for next substitution

        if len(substitutions) == 0:                                                 # Check if substitution list is empty
            print("No substitutions observed")                                      # Explain why table has no rows

    return sample_summary                                                          # Return completed dictionary
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,
        "control_n": 3,
        "treated_n": 3,
        "control_mean": 0.5,
        "treated_mean": 2.0,
        "difference": 1.5,
        "statistic": ...,
        "p_value": ...
    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.


# Step 5 - Compare mismatch positions between groups

def compare_positions(sample_summary, control_group="control", treated_group="treatment"):  # Define function
    combined_counts = {}                                                                    # Create dictionary for total mismatch counts
    results = []                                                                            # Create list for statistical results
    excluded_samples = []                                                                   # Create list for samples with no valid reads

    for sample in sample_summary:                                                           # Loop through each sample
        group = sample_summary[sample]["group"]                                             # Get the group name

        if group == control_group or group == treated_group:                                # Check if sample belongs to either group
            position_counts = sample_summary[sample]["position_counts"]                     # Get mismatch positions

            for position in position_counts:                                                # Loop through each mismatch position
                if position not in combined_counts:                                         # Check if position is new
                    combined_counts[position] = 0                                           # Start count at zero

                combined_counts[position] += position_counts[position]                      # Add mismatches at this position

            if sample_summary[sample]["valid_reads"] == 0:                                  # Check for samples with no valid reads
                excluded_samples.append(sample)                                             # Record sample for exclusion

    ranked_positions = sorted(combined_counts)                                              # Sort positions from lowest to highest

    ranked_positions = sorted(                                                              # Sort positions by mismatch frequency
        ranked_positions,                                                                   # Use the previously sorted positions
        key=lambda position: combined_counts[position],                                     # Use mismatch count for sorting
        reverse=True                                                                        # Put highest mismatch counts first
    )

    top_positions = ranked_positions[:3]                                                    # Select three most frequent positions

    for sample in excluded_samples:                                                         # Loop through excluded samples
        print(f"Excluded {sample}: no valid reads")                                         # Report why sample was excluded

    for position in top_positions:                                                          # Loop through selected positions
        control_percentages = []                                                            # Create list of control sample percentages
        treated_percentages = []                                                            # Create list of treatment sample percentages

        for sample in sample_summary:                                                       # Loop through each biological sample
            group = sample_summary[sample]["group"]                                         # Get sample group
            valid_reads = sample_summary[sample]["valid_reads"]                             # Get number of valid reads
            position_counts = sample_summary[sample]["position_counts"]                     # Get position counts

            if valid_reads > 0:                                                             # Exclude samples without valid reads
                if position in position_counts:                                             # Check if selected position has mismatches
                    count = position_counts[position]                                       # Get mismatch count at this position
                else:                                                                       # If position has no mismatches
                    count = 0                                                               # Include zero mismatches

                percentage = count / valid_reads * 100                                      # Calculate mismatch percentage

                if group == control_group:                                                  # Check if sample is a control
                    control_percentages.append(percentage)                                  # Add control sample percentage

                if group == treated_group:                                                  # Check if sample received treatment
                    treated_percentages.append(percentage)                                  # Add treated sample percentage

        control_n = len(control_percentages)                                                # Count control biological samples
        treated_n = len(treated_percentages)                                                # Count treated biological samples

        if control_n > 0:                                                                   # Check if controls are available
            control_mean = sum(control_percentages) / control_n                             # Calculate mean control percentage
        else:                                                                               # If no valid control samples
            control_mean = None                                                             # Mean cannot be calculated

        if treated_n > 0:                                                                   # Check if treated samples are available
            treated_mean = sum(treated_percentages) / treated_n                             # Calculate mean treated percentage
        else:                                                                               # If no valid treated samples
            treated_mean = None                                                             # Mean cannot be calculated

        if control_mean != None and treated_mean != None:                                   # Check if both means are available
            difference = treated_mean - control_mean                                        # Calculate difference in percentage points
        else:                                                                               # If either mean is unavailable
            difference = None                                                               # Difference cannot be calculated

        statistical_result = calculate_statistics(                                          # Run supplied Welch's t-test function
            control_percentages,                                                            # Pass control sample percentages
            treated_percentages                                                             # Pass treated sample percentages
        )

        statistic = statistical_result["statistic"]                                         # Get t-statistic
        p_value = statistical_result["p_value"]                                             # Get p-value

        results.append({                                                                    # Add results for this position
            "position": position,                                                           # Store DNA position
            "control_n": control_n,                                                         # Store number of control samples
            "treated_n": treated_n,                                                         # Store number of treated samples
            "control_mean": control_mean,                                                   # Store control mean percentage
            "treated_mean": treated_mean,                                                   # Store treated mean percentage
            "difference": difference,                                                       # Store difference between group means
            "statistic": statistic,                                                         # Store Welch's t-statistic
            "p_value": p_value                                                              # Store statistical p-value
        })

        print(
            f"Position {position}: "                                                        # Show the mismatch position
            f"Control n={control_n}, mean={control_mean}; "                                 # Show number of control samples and their mean
            f"Treatment n={treated_n}, mean={treated_mean}; "                               # Show number of treated samples and their mean
            f"Difference={difference}; "                                                    # Show the difference between group means
            f"t={statistic}, p={p_value}"                                                   # Show the t-statistic and p-value
        )

    return results                                                                          # Return list of statistical comparisons
````
